```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt
```

```python
urls = ["https://images.alodokter.com/dk0z4ums3/image/upload/v1773189711/attached_image/kucing-bengal-inilah-ciri-khas-karakter-dan-cara-merawatnya.jpg", "https://dansmedia.net/wp-content/uploads/2018/11/Foto-Utama.jpg" ,"https://d1vbn70lmn1nqe.cloudfront.net/prod/wp-content/uploads/2025/08/04114156/kucing-oren.jpg"]

for url in urls :
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image,image_2))
  cv2_imshow(final_frame)
  print('\n')
```

# image.png

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar= ", tinggi , " x " , lebar)
cv2_imshow(image_2)
```

```python
#Buat Gars vertikal
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_4 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range (tinggi) :
  image_4[x, int(lebar/2)] = [255,255,255]

final_frame = cv.hconcat((image_2,image_4))
cv2_imshow(final_frame)
```

image.png

```python
# buat garis menyilang diagonal pada image
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_5 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(tinggi):
    x = int((lebar / tinggi) * y)
    image_5[y, x] = [255, 255, 255]

for y in range(tinggi):
    x = lebar - 1 - int((lebar / tinggi) * y)
    image_5[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_5))
cv2_imshow(final_frame)
```

image.png

```python
#garis horizontal putih di tengah gambar
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_6 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tengah_x = int(lebar / 2)
tengah_y = int(tinggi / 2)

for x in range(tengah_x - 25, tengah_x + 25):
    image_6[tengah_y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_6))
cv2_imshow(final_frame)
```

```python
# kotak menggunakan kumpulan pixel warna putih
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_7 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

tengah_x = int(lebar / 2)
tengah_y = int(tinggi / 2)

for y in range(tengah_x - 25, tengah_x + 25):
    for x in range(tengah_y - 25, tengah_y + 25):
        image_7[x, y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_7))

cv2_imshow(final_frame)
```
